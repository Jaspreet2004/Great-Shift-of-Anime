```python
import pandas as pd
import matplotlib.pyplot as plt
import ast
from collections import Counter

df = pd.read_csv("top_1000_anime.csv")
print("Loaded dataset with", len(df), "rows")


df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year"])
df["year"] = df["year"].astype(int)

df_1 = df[(df["year"] >= 1970) & (df["year"] < 2000)]
df_2 = df[df["year"] >= 2000]

#First graph 1970-1999
year_counts_1 = df_1["year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
year_counts_1.plot(kind="bar")

plt.title("Top 1000 Anime by Year (1970–1999)")
plt.xlabel("Year")
plt.ylabel("Count")

years_1 = year_counts_1.index
plt.xticks(
    ticks=range(0, len(years_1), 3),
    labels=years_1[::3],
    rotation=45
)

plt.tight_layout()
plt.show()

#Graph 2 2000-Present
year_counts_2 = df_2["year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
year_counts_2.plot(kind="bar")

plt.title("Top 1000 Anime by Year (2000–Present)")
plt.xlabel("Year")
plt.ylabel("Count")

years_2 = year_counts_2.index
plt.xticks(
    ticks=range(0, len(years_2), 3),
    labels=years_2[::3],
    rotation=45
)

plt.tight_layout()
plt.show()

#Genre graph
all_genres = []

for genres_str in df["genres_raw"]:
    try:
        genres_list = ast.literal_eval(genres_str)
        for g in genres_list:
            if isinstance(g, dict) and "name" in g:
                all_genres.append(g["name"])
    except:
        continue

genre_counts = Counter(all_genres)
top_genres = dict(genre_counts.most_common(12))

#Genre graph
plt.figure(figsize=(10, 5))
plt.bar(top_genres.keys(), top_genres.values())

plt.title("Top 12 Most Common Genres in Top 1000 Anime")
plt.xlabel("Genre")
plt.ylabel("Count")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
